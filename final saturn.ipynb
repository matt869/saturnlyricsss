```python
import time
import sys

def type_with_speed(text, speed=0.1, pause_positions=None):
    
    if pause_positions is None:
        pause_positions = []
    for i, char in enumerate(text):
        sys.stdout.write(char)
        sys.stdout.flush()
        if i in pause_positions:
            time.sleep(2.5)  
        else:
            time.sleep(speed)
    print()

# Lyrics 
lyrics = [
    ("Ooh (Ooh, ooh)", 0.2, [2]), 
    ("Life's better on Saturn", 0.1),
    ("Got to break this pattern", 0.1),
    ("Of floating away", 0.1), 
    ("Ooh (Ooh, ooh)", 0.2,), 
    ("", 0), 
    ("Find something worth saving", 0.1), 
    ("", 0, [0]),  
    ("It's all for the taking", 0.1), 
    ("I always say", 0.1) 
]


for line_info in lyrics:
    if len(line_info) == 3:
        text, speed, pause_positions = line_info
    elif len(line_info) == 2:
        text, speed = line_info
        pause_positions = []
    if text:
        type_with_speed(text, speed, pause_positions)
```
